```python
import pandas as pd
```

```python
df = pd.read_csv("../data/steam.csv")

print("=" * 50)
print("SHAPE & STRUCTURE")
print("=" * 50)
print("Rows, Columns:", df.shape)
print("\nDtypes:\n", df.dtypes)

print("\n" + "=" * 50)
print("MISSING VALUES")
print("=" * 50)
missing = df.isnull().sum()
print(missing[missing > 0] if missing.sum() > 0 else "No nulls found.")

print("\n" + "=" * 50)
print("DUPLICATES")
print("=" * 50)
print("Full duplicate rows:", df.duplicated().sum())
print("Duplicate appid (should be 0 — appid is the primary key):", df['appid'].duplicated().sum())

print("\n" + "=" * 50)
print("NUMERIC SUMMARY (price, reviews, playtime)")
print("=" * 50)
print(df[['price', 'positive_ratings', 'negative_ratings',
          'average_playtime', 'median_playtime']].describe())

print("\n" + "=" * 50)
print("STRUCTURAL FLAGS — multi-value fields")
print("=" * 50)
for col in ['platforms', 'categories', 'genres', 'steamspy_tags']:
    print(f"\n{col} — sample values:")
    print(df[col].dropna().head(3).tolist())
```

```python
# Structural sanity checks only — not business-logic cleaning
assert df['appid'].is_unique, "appid should be unique — investigate before proceeding"
assert df.dropna(how='all').shape[0] == df.shape[0], "Found fully-empty rows"

# Export exactly as received — this is what gets uploaded to Databricks as Bronze
df.to_csv("../data/steam_bronze.csv", index=False)
print(f"\nExported {df.shape[0]} rows, {df.shape[1]} columns → steam_bronze.csv")
```
